
# Introduction to NumPy 



## 1) Note the difference between Python **lists** and NumPy **arrays**.

In Python, we can store collections using **lists**. NumPy adds **arrays**, which are faster and more powerful for math. Run the cells to see the difference between a list and a NumPy array.


```python
# Import NumPy and check version
import numpy as np
np.__version__
```

```python
# Lists vs Arrays
py_list = [1, 2, 3]
np_array = np.array([1, 2, 3])

print('Python list type:', type(py_list))
print('NumPy array type:', type(np_array))
print('Array + 10 (elementwise):', np_array + 10)  # works elementwise
print(py_list + 10)  # gives error; lists don't do elementwise math
```


## 2) Creating Arrays


- `np.array([...])`: from a Python list  
- `np.zeros(shape)`, `np.ones(shape)`, `np.full(shape, value)`: filled arrays  
- `np.eye(n)`: identity matrix (1s on the diagonal)  
- `np.arange(start, stop, step)`: numbers in a range (**stop is not included**)  
- `np.linspace(start, stop, num)`: `num` evenly spaced values (**including** `stop`)  
- Random arrays for practice: use a **generator** `rng = np.random.default_rng(seed)`


```python
a = np.array([10, 20, 30])
z = np.zeros((2, 3))  # shape (2, 3)
o = np.ones((3, 2))   # shape (3, 2)
f = np.full((2, 2), 7)
I = np.eye(4)

print('a:', a)
print('\n z:', z)
print('\n o:', o)
print('\n z shape:', z.shape, 'o shape:', o.shape)
print('\n f:', f)
print('\n I:', I)
```

```python
r1 = np.arange(0, 10, 2)     # 0,2,4,6,8
r2 = np.linspace(0, 1, 5)    # 0.00, 0.25, 0.50, 0.75, 1.00
print('arange:', r1)
print('linspace:', r2)
```

```python
# Random numbers (reproducible)
rng = np.random.default_rng(42)
rng.random((2, 3))  # uniform in [0, 1)
```


## 3) Inspecting Arrays

- `.ndim`: number of dimensions (1D, 2D, …)  
- `.shape`: size along each dimension (rows, columns)  
- `.size`: total number of elements  
- `.dtype`: data type (e.g., `int64`, `float64`)


```python
X = np.arange(12).reshape(3, 4)  # 3 rows, 4 columns
print('X:', X)
print('ndim:', X.ndim, ', shape:', X.shape, ', size:', X.size, ', dtype:', X.dtype)
```


## 4) Indexing & Slicing  

- `a[0]` → first element (Python is 0‑indexed)  
- `a[1:5]` → elements from index 1 **up to, not including** 5  
- `A[row, col]` → 2D indexing  
- `A[:, 0]` → all rows, first column  
- `A[::2]` → every 2nd element  


```python
a = np.arange(10)
print('a:', a)
print('a[0]:', a[0])
print('a[1:5]:', a[1:5])
print('a[::-1]:', a[::-1])  # reverse
```

```python
A = np.arange(12).reshape(3, 4)
print('A:', A)
print('A[1, 2]:', A[1, 2])     # row 1, col 2
print('A[:, 0]:', A[:, 0])     # first column
print('A[::2, 1:]:', A[::2, 1:])  # every 2nd row, columns 1..end
```


## 5) Vectorized Math & Simple Statistics


NumPy applies operations to **all elements at once** (vectorization):
- Elementwise operators: `+`, `-`, `*`, `/`, `**`  
- Functions: `np.abs`, `np.sqrt`, `np.log`  
- Aggregations: `mean`, `sum`, `min`, `max`, `argmax`  

**Axis** controls the direction of a reduction:
- `X.mean(axis=0)` → mean of each **column**  
- `X.mean(axis=1)` → mean of each **row**  

> **Missing values (NaN):** If data has `np.nan`, regular means become `nan`. Use `np.nanmean(X)` to safely ignore NaNs.


```python
x = np.array([0., 1., 2., 3.])
y = (x + 1)**2        # no loop!
print('y:', y)              # [1, 4, 9, 16]
print('sqrt(y):', np.sqrt(y))
```

```python
rng = np.random.default_rng(42)
X = rng.random((5, 3))
print('X:', X)
print('column means:', X.mean(axis=0))
print('row minima:', X.min(axis=1))
print('overall max value:', X.max(), 'at flat index:', X.argmax())
```


## 6) Reshape & Combine


- `reshape(new_shape)` rearranges elements (same total size).  
- `ravel()` returns a flattened 1D **view** if possible.  
- Combine arrays:  
  - `np.hstack([A, B])` → stack **columns** (side‑by‑side)  
  - `np.vstack([A, B])` → stack **rows** (one on top of another)  
  - `np.concatenate([...], axis=0 or 1)` → general version


```python
v = np.arange(12)        # 0..11
M = v.reshape(3, 4)      # 3 rows, 4 columns
print('M:', M)
print('flattened:', M.ravel())
```

```python
A = np.ones((2, 3))
B = np.zeros((2, 3))
print('hstack:', np.hstack([A, B]))  # (2, 6)
print('vstack:', np.vstack([A, B]))  # (4, 3)
print('concatenate (axis=0):',np.concatenate((A, B),axis=0))
print('concatenate (axis=1):',np.concatenate((A, B),axis=1))
```


## 7) Randomness & Reproducibility

- `rng = np.random.default_rng(42)` (42 is the seed)  
- `rng.random(shape)`
- `rng.integers(low, high, size)` 
- `rng.normal(mean, std, size)`


```python
rng = np.random.default_rng(42)
print('Random 2x2:', rng.random((2, 2)))
print('Integers:', rng.integers(0, 10, size=(2, 3)))
print('Numbers from a normal distribution:',rng.normal(0, 1, (2,2)))
```
