```python
def Josephus(person_array, index):

  # Base case, when only k-1 persons are left
  if len(person_array) == k-1:
    return
   
  # Finding the index of first person who will be removed
  index = ((index+(k-1))%len(person_array))
   
  # Removing the person
  person_array.pop(index)

  # Printing the array after removal
  print(person_array)
   
  # Recursive call for n-1 persons
  Josephus(person_array,index)
```

```python
# Specific n and k for Josephus problem

n = 10  # Number of persons 
k = 2   # Every  kth person is killed


index = 0  # Python indexing starts from 0
 
    
# Fill the person array
person_array=[]
for i in range(1,n+1):
  person_array.append(i)


# Function call
Josephus(person_array,index)


print("Survivor(s):", ", ".join(map(str, person_array)))
```
