```python
import time
```

```python
MOVE_NO = 0   # counts moves from 1 to (2^n - 1)

def hanoi(n, source, auxiliary, destination):
    """
    Move n disks from 'source' to 'destination' using 'auxiliary'.
    Standard recursive solution:
      1) Move n-1 disks from 'source' to 'auxiliary'
      2) Move largest disk from 'source' to 'destination'
      3) Move n-1 disks from 'auxiliary' to 'destination'
    """
    global MOVE_NO, N

    # Base case: only one disk to move
    if n == 1:
        #MOVE_NO += 1
        print(f"Move disk 1 from {source} to {destination}")
        #print(f"Move {MOVE_NO} ({MOVE_NO:0{N}b}): disk 1 from {source} to {destination}")
        return

    # Step 1: move n-1 disks to auxiliary
    hanoi(n - 1, source, destination, auxiliary)

    # Step 2: move the largest (nth) disk to destination
    #MOVE_NO += 1
    print(f"Move disk {n} from {source} to {destination}")
    #print(f"Move {MOVE_NO} ({MOVE_NO:0{N}b}): disk {n} from {source} to {destination}")

    # Step 3: move n-1 disks from auxiliary to destination
    hanoi(n - 1, auxiliary, source, destination)
```

```python
# Input for number of disks
N = 4

print("\nSequence of moves:")
start_time = time.time()
hanoi(N, "A", "B", "C")
end_time = time.time()

#print(f"\nTotal moves: {MOVE_NO} (expected {2**N - 1})")
print("\nTime taken: ",end_time-start_time," sec")
```
